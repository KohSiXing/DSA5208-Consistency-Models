# Connections

```python
from pymongo import MongoClient
from pymongo.write_concern import WriteConcern
from pymongo.read_concern import ReadConcern

# Normal replica-set connection.
# Writes will normally go to the primary.
client = MongoClient(
    "mongodb://mongo1:27017,mongo2:27017,mongo3:27017/"
    "?replicaSet=dsa5208",
    serverSelectionTimeoutMS=5000,
)

# Direct connections to individual servers.
client2 = MongoClient(
    "mongodb://mongo2:27017/?directConnection=true",
    serverSelectionTimeoutMS=5000,
)

client3 = MongoClient(
    "mongodb://mongo3:27017/?directConnection=true",
    serverSelectionTimeoutMS=5000,
)
```

# Check Replica-Set Status

Before running the experiment, confirm that MongoDB can see the replica set.

```python
hello = client.admin.command("hello")

print("Replica set:", hello.get("setName"))
print("Primary:", hello.get("primary"))
print("Writable primary:", hello.get("isWritablePrimary"))

status = client.admin.command("replSetGetStatus")

for member in status["members"]:
    print(member["name"], member["stateStr"])
```

# Database and Collection

The database is called `university` and the collection is called `students`.

MongoDB creates them automatically when we insert the first document.

```python
def get_collections(write_concern_value, read_concern_value):
    """
    Returns two collection objects:

    writes:
        Uses the replica-set connection.
        Writes go to the primary.

    reads:
        Uses the direct mongo3 connection.
        Reads come specifically from mongo3.
    """

    write_db = client.get_database(
        "university",
        write_concern=WriteConcern(w=write_concern_value)
    )

    read_db = client3.get_database(
        "university",
        read_concern=ReadConcern(level=read_concern_value)
    )

    return write_db["students"], read_db["students"]
```

# Control Mongo3 Replication

We pause mongo3 so that it deliberately falls behind the other servers.

```python
def stall_mongo3():
    """Pause replication on mongo3."""
    client3.admin.command({
        "configureFailPoint": "rsSyncApplyStop",
        "mode": "alwaysOn"
    })


def unstall_mongo3():
    """Resume replication on mongo3."""
    client3.admin.command({
        "configureFailPoint": "rsSyncApplyStop",
        "mode": "off"
    })
```

# Reusable Experiment Functions

These functions create data, update data, and read data from mongo3.

```python
def prepare_document(writes, student_id, initial_score):
    """Create or reset the test document."""
    writes.update_one(
        {"student_id": student_id},
        {
            "$set": {
                "student_id": student_id,
                "score": initial_score
            }
        },
        upsert=True
    )
    print(f"Initial score for {student_id}: {initial_score}")


def update_score(writes, student_id, new_score):
    """Update the score using the selected write concern."""
    writes.update_one(
        {"student_id": student_id},
        {"$set": {"score": new_score}}
    )
    print(f"New score written: {new_score}")


def read_from_mongo3(reads, student_id):
    """Read and print the document specifically from mongo3."""

    try:
        result = reads.find_one(
            {"student_id": student_id},
            max_time_ms=3000
        )

        if result is None:
            print("No document found.")
            return None

        print("Document read from mongo3:")
        print(f"  student_id: {result.get('student_id')}")
        print(f"  score:      {result.get('score')}")

        return result

    except Exception as error:
        print("Read failed or timed out.")
        print(f"Error: {error}")
        return None

    except Exception as error:
        print("Read failed or timed out:", error)
        return None
```

## Experiment: Write Concern 1 / Read Concern Majority

The configuration used was:

- **Write concern:** `w = 1`
- **Read concern:** `majority`
- **Database:** `university`
- **Collection:** `students`

The test document initially contained `student_id: TEST_1_MAJORITY` and `score: 0`.

### Prediction

For the `1 / Majority` configuration, we predict that read-your-writes consistency may not be guaranteed.

The write concern `w = 1` allows the primary to acknowledge the write after only one server confirms it. Therefore, the write may be acknowledged before the update reaches all secondary servers.

Because replication on `mongo3` is paused, mongo3 may still contain the older value `score = 0` when the client performs its read. The client may therefore observe:

- Client writes `score = 1`
- Client reads `score = 0`



```python
def test_one_majority():
    student_id = "TEST_1_MAJORITY"

    # Select Write concern 1 and Read concern majority.
    writes, reads = get_collections(1, "majority")

    # Prepare the initial value while all nodes are healthy.
    prepare_document(writes, student_id, 0)

    # Pause mongo3 replication.
    stall_mongo3()

    try:
        # Write score = 1 using w=1.
        update_score(writes, student_id, 1)

        # Read score from mongo3 using majority read concern.
        read_from_mongo3(reads, student_id)

    finally:
        # Always restart mongo3.
        unstall_mongo3()
```

```python
test_one_majority()
```

### Interpretation

Client wrote `score = 1`, but a read from the lagging `mongo3` returned the old value `0` — because mongo3 hadn't yet received the update. This violates **read-your-writes consistency** (a later read by the same client must never return an older value than what it just wrote). **Conclusion:** under `w=1 / readConcern=majority`, read-your-writes was **not guaranteed** while mongo3's replication was delayed.

## Experiment: Write Concern Majority / Read Concern Local

The configuration used was:

- **Write concern:** `w = "majority"`
- **Read concern:** `local`
- **Database:** `university`
- **Collection:** `students`

The test document initially contained `student_id: TEST_MAJORITY_LOCAL` and `score: 0`.

### Prediction

For the `Majority / Local` configuration, we predict that read-your-writes consistency may not be guaranteed.

The write concern `w = "majority"` requires the write to be acknowledged by most of the replica-set members. However, the read concern `local` allows the read to return the value currently stored on the selected server, even if that server is behind.

Because replication on `mongo3` is paused, mongo3 may still contain the older value `score = 0` when the client performs its read. The client may therefore observe:

- Client writes `score = 1`
- Client reads `score = 0`

If this occurs, read-your-writes consistency is violated because the client reads an older value after its own write.

```python
def test_majority_local():
    student_id = "TEST_MAJORITY_LOCAL"

    # Select Write concern majority and Read concern local.
    writes, reads = get_collections("majority", "local")

    # Prepare the initial value while all nodes are healthy.
    prepare_document(writes, student_id, 0)

    # Pause mongo3 replication.
    stall_mongo3()

    try:
        # Write score = 1 using majority write concern.
        update_score(writes, student_id, 1)

        # Read score from mongo3 using local read concern.
        read_from_mongo3(reads, student_id)

    finally:
        # Always restart mongo3.
        unstall_mongo3()
```

```python
test_majority_local()
```

### Interpretation

Client wrote `score = 1`, but a read from the lagging `mongo3` returned the old value `0` because mongo3 had not yet received the update. This violates **read-your-writes consistency**, since a later read by the same client must not return an older value than the value it just wrote.

**Conclusion:** under `w="majority" / readConcern="local"`, read-your-writes was **not guaranteed** while mongo3's replication was delayed.
